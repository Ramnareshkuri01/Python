```python
import random
import numpy as np

# 1. Generate a dataset of 10 random 0s and 1s
# We'll assume an underlying probability of 0.5 for demonstration, but it could be any p.
# random.choices allows for sampling with replacement from a population.
dataset_bernoulli = random.choices([0, 1], k=100)

print(f"Generated Bernoulli dataset: {dataset_bernoulli}")

# 2. 'Run Bernoulli on that' - Calculate the empirical probability of success (p_hat)
p_hat = sum(dataset_bernoulli) / len(dataset_bernoulli)

print(f"Empirical probability of success (p_hat) from the dataset: {p_hat:.2f}")

# You can also use numpy for this calculation for larger datasets
# p_hat_np = np.mean(dataset_bernoulli)
# print(f"Empirical probability (using numpy): {p_hat_np:.2f}")
```

```python


# Number of trials (n)
n = len(dataset_bernoulli)
# Number of successes (k) - sum of 1s in the dataset
k = sum(dataset_bernoulli)

# Define the Bernoulli likelihood function
def bernoulli_likelihood(p, k, n):
    # Clamp p to avoid log(0) or division by zero issues for probabilities very close to 0 or 1
    p = np.clip(p, 1e-10, 1 - 1e-10)
    # Likelihood = p^k * (1-p)^(n-k)
    return (p**k) * ((1-p)**(n-k))

# Create a range of p values from 0 to 1 for plotting
p_values = np.linspace(0, 1, 1000)

# Calculate the likelihood for each p_value
likelihoods = [bernoulli_likelihood(p, k, n) for p in p_values]

# Plot the likelihood function
plt.figure(figsize=(10, 6))
plt.plot(p_values, likelihoods, label=f'Likelihood function for k={k}, n={n}')

# Calculate and highlight the Maximum Likelihood Estimate (MLE)
mle_p = k / n
mle_likelihood = bernoulli_likelihood(mle_p, k, n)
plt.axvline(x=mle_p, color='r', linestyle='--', label=f'MLE of p: {mle_p:.2f}')
plt.scatter(mle_p, mle_likelihood, color='red', zorder=5, s=100, label='Max Likelihood Point')

plt.title('Bernoulli Likelihood Function')
plt.xlabel('Probability of Success (p)')
plt.ylabel('Likelihood L(p|D)')
plt.grid(True)
plt.legend()
plt.show()
```

This code defines the `bernoulli_likelihood` function and then plots its values across the possible range of `p` (0 to 1), highlighting the Maximum Likelihood Estimate (MLE) which is the `p` that maximizes the likelihood of your observed data.
