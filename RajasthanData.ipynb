```python
import pandas as pd

df = pd.read_csv('data-1744177001524 (2).csv')
df.head()  # Show the first few rows


```

```python
# prompt: how many districts are there

num_districts = len(df['district_name'].unique())
print(f"There are {num_districts} districts.")
```

```python

# Group by district and count the number of villages
district_village_counts = df.groupby('district_name')['village_name'].count()

# Print the results
district_village_counts
```

```python
# prompt:  all districts with number of villages in graph,make this graph interactive
from google.colab import files
import pandas as pd
import plotly.express as px

# Group by district and count the number of villages
district_village_counts = df.groupby('district_name')['village_name'].count().reset_index()
district_village_counts.rename(columns={'village_name': 'village_count'}, inplace=True)

# Create an interactive bar chart using Plotly
fig = px.bar(district_village_counts, x='district_name', y='village_count',
             labels={'district_name': 'District', 'village_count': 'Number of Villages'},
             title='Number of Villages per District',
             hover_data=['district_name', 'village_count'])

fig.update_layout(xaxis_title='District', yaxis_title='Number of Villages')
fig.show()
```

```python
# prompt: top 10 districts with highest number of villages

# Sort the districts by village count in descending order and get the top 10
top_10_districts = district_village_counts.sort_values(by='village_count', ascending=False).head(10)

# Display the top 10 districts
print(top_10_districts)

# Create an interactive bar chart for the top 10 districts
fig_top_10 = px.bar(top_10_districts, x='district_name', y='village_count',
                     labels={'district_name': 'District', 'village_count': 'Number of Villages'},
                     title='Top 10 Districts with Highest Number of Villages',
                     hover_data=['district_name', 'village_count'])

fig_top_10.update_layout(xaxis_title='District', yaxis_title='Number of Villages')
fig_top_10.show()
```

```python
# prompt: lowest 10 districts with lowest number of villages

# Sort the districts by village count in ascending order and get the lowest 10
lowest_10_districts = district_village_counts.sort_values(by='village_count', ascending=True).head(10)

# Display the lowest 10 districts
print(lowest_10_districts)

# Create an interactive bar chart for the lowest 10 districts
fig_lowest_10 = px.bar(lowest_10_districts, x='district_name', y='village_count',
                     labels={'district_name': 'District', 'village_count': 'Number of Villages'},
                     title='Lowest 10 Districts with Lowest Number of Villages',
                     hover_data=['district_name', 'village_count'])

fig_lowest_10.update_layout(xaxis_title='District', yaxis_title='Number of Villages')
fig_lowest_10.show()
```

```python
# prompt: create top 5 districts with highest and lowest literacy rate in graph use Male Literacy Rate: male_literacy / male_literacy_rate_denom * 100
# Female Literacy Rate: female_literacy / female_literacy_rate_denom * 100

# Calculate Male and Female Literacy Rates
df['male_literacy_rate'] = df['male_literacy'] / df['male_literacy_rate_denom'] * 100
df['female_literacy_rate'] = df['female_literacy'] / df['female_literacy_rate_denom'] * 100

# Group data by district and calculate the average literacy rates
district_literacy = df.groupby('district_name').agg({'male_literacy_rate': 'mean', 'female_literacy_rate': 'mean'}).reset_index()

# Find top 5 districts with highest literacy rate
top_5_highest_literacy = district_literacy.sort_values(by='male_literacy_rate', ascending=False).head(5)

# Find top 5 districts with lowest literacy rate
top_5_lowest_literacy = district_literacy.sort_values(by='male_literacy_rate', ascending=True).head(5)

# Create the plot for highest literacy rates
fig_highest = px.bar(top_5_highest_literacy, x='district_name', y='male_literacy_rate',
                     labels={'district_name': 'District', 'male_literacy_rate': 'Male Literacy Rate (%)'},
                     title='Top 5 Districts with Highest Male Literacy Rates',
                     color='male_literacy_rate')
fig_highest.show()

# Create the plot for lowest literacy rates
fig_lowest = px.bar(top_5_lowest_literacy, x='district_name', y='male_literacy_rate',
                     labels={'district_name': 'District', 'male_literacy_rate': 'Male Literacy Rate (%)'},
                     title='Top 5 Districts with Lowest Male Literacy Rates',
                     color='male_literacy_rate')
fig_lowest.show()
```

```python
# prompt: top 5 and bottom 5 districts in women literacy rate

# Find top 5 districts with highest female literacy rate
top_5_highest_female_literacy = district_literacy.sort_values(by='female_literacy_rate', ascending=False).head(5)

# Find bottom 5 districts with lowest female literacy rate
bottom_5_lowest_female_literacy = district_literacy.sort_values(by='female_literacy_rate', ascending=True).head(5)

# Create the plot for highest female literacy rates
fig_highest_female = px.bar(top_5_highest_female_literacy, x='district_name', y='female_literacy_rate',
                     labels={'district_name': 'District', 'female_literacy_rate': 'Female Literacy Rate (%)'},
                     title='Top 5 Districts with Highest Female Literacy Rates',
                     color='female_literacy_rate')
fig_highest_female.show()

# Create the plot for lowest female literacy rates
fig_lowest_female = px.bar(bottom_5_lowest_female_literacy, x='district_name', y='female_literacy_rate',
                     labels={'district_name': 'District', 'female_literacy_rate': 'Female Literacy Rate (%)'},
                     title='Bottom 5 Districts with Lowest Female Literacy Rates',
                     color='female_literacy_rate')
fig_lowest_female.show()
```

```python
# prompt: top 5 villages of highest female literacy rate show in graph

# Sort the DataFrame by female literacy rate in descending order and get the top 5
top_5_villages = df.sort_values(by='female_literacy_rate', ascending=False).head(5)

# Create the plot for the top 5 villages with highest female literacy rates
fig_top_5_villages = px.bar(top_5_villages, x='village_name', y='female_literacy_rate',
                           labels={'village_name': 'Village', 'female_literacy_rate': 'Female Literacy Rate (%)'},
                           title='Top 5 Villages with Highest Female Literacy Rates',
                           color='female_literacy_rate')
fig_top_5_villages.show()
```

```python
# prompt: bottom 5 villages of lowest female literacy rate show in graph

# Sort the DataFrame by female literacy rate in descending order and get the top 5
top_5_villages = df.sort_values(by='female_literacy_rate', ascending=True).head(10)

# Create the plot for the top 5 villages with highest female literacy rates
fig_bottom_5_villages = px.bar(top_5_villages, x='village_name', y='female_literacy_rate',
                           labels={'village_name': 'Village', 'female_literacy_rate': 'Female Literacy Rate (%)'},
                           title='Top 5 Villages with Highest Female Literacy Rates',
                           color='female_literacy_rate')
fig_bottom_5_villages.show()
```

```python
# prompt: total tubewells district wise

# Calculate total tubewells per district
district_tubewell_counts = df.groupby('district_name')['no_of_tubewells'].sum().reset_index()

# Rename the column for clarity
district_tubewell_counts.rename(columns={'no_of_tubewells': 'total_tubewells'}, inplace=True)

# Display the results
print(district_tubewell_counts)

# Create an interactive bar chart for the total tubewells per district
fig_tubewells = px.bar(district_tubewell_counts, x='district_name', y='total_tubewells',
                       labels={'district_name': 'District', 'total_tubewells': 'Total Tubewells'},
                       title='Total Tubewells per District',
                       hover_data=['district_name', 'total_tubewells'])

fig_tubewells.update_layout(xaxis_title='District', yaxis_title='Total Tubewells')
fig_tubewells.show()
```
