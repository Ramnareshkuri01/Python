## Visualize and Check Stationarity (Rajasthan GDP)

### Subtask:
Plot Rajasthan's historical GDP to observe its trend and check for stationarity using the Augmented Dickey-Fuller (ADF) statistical test.


### Plotting Rajasthan's GDP and Performing ADF Test

To visualize the historical trend of Rajasthan's GDP, a line plot will be generated. This plot will help in understanding the general trajectory of economic growth over time. Following the visualization, the Augmented Dickey-Fuller (ADF) test will be performed. This statistical test is crucial for determining if the time series data is stationary, a fundamental assumption for many time series forecasting models like ARIMA. The test results, including the test statistic, p-value, and critical values, will indicate whether the null hypothesis of non-stationarity can be rejected.

**Reasoning**:
I will now generate a line plot for Rajasthan's historical GDP using the 'gdp' column from the `rajasthan_gdp_df` DataFrame, setting the datetime index as the x-axis. After visualizing the trend, I will perform the Augmented Dickey-Fuller (ADF) test on the 'gdp' column to check for stationarity and print its results, which are crucial for time series analysis.



```python
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.stattools import adfuller

# 1. Plot Rajasthan's historical GDP
plt.figure(figsize=(14, 7))
sns.lineplot(data=rajasthan_gdp_df, x=rajasthan_gdp_df.index, y='gdp', marker='o')
plt.title("Rajasthan's Historical GDP Trend")
plt.xlabel('Year')
plt.ylabel('Real GDP')
plt.grid(True)
plt.show()

# 2. Perform the Augmented Dickey-Fuller test
print("\nPerforming Augmented Dickey-Fuller test on Rajasthan's GDP:")
adf_result = adfuller(rajasthan_gdp_df['gdp'])

print(f'ADF Statistic: {adf_result[0]}')
print(f'p-value: {adf_result[1]}')
print('Critical Values:')
for key, value in adf_result[4].items():
    print(f'   {key}: {value}')

if adf_result[1] <= 0.05:
    print("Conclusion: The time series is likely stationary (reject H0).")
else:
    print("Conclusion: The time series is likely non-stationary (fail to reject H0).")
```

## Determine ARIMA Orders (p, d, q)

### Subtask:
Based on the stationarity test, determine the 'd' (differencing) order. Then, use Autocorrelation Function (ACF) and Partial Autocorrelation Function (PACF) plots to help identify appropriate 'p' (AR order) and 'q' (MA order) values for the ARIMA model.


**Reasoning**:
Since the previous ADF test indicated non-stationarity for the original GDP series, the first step to determine the 'd' (differencing) order for the ARIMA model is to apply first-order differencing. I will calculate the difference of the 'gdp' column and store it in a new column called 'gdp_diff'.



```python
rajasthan_gdp_df['gdp_diff'] = rajasthan_gdp_df['gdp'].diff().dropna()
print(rajasthan_gdp_df.head())
```

**Reasoning**:
Now that first-order differencing has been applied, I will perform the Augmented Dickey-Fuller (ADF) test on the `gdp_diff` column to confirm if the series is now stationary. The results of this test will help determine the 'd' order for the ARIMA model.



```python
from statsmodels.tsa.stattools import adfuller

print("\nPerforming Augmented Dickey-Fuller test on Rajasthan's differenced GDP (gdp_diff):")
adf_diff_result = adfuller(rajasthan_gdp_df['gdp_diff'].dropna())

print(f'ADF Statistic: {adf_diff_result[0]}')
print(f'p-value: {adf_diff_result[1]}')
print('Critical Values:')
for key, value in adf_diff_result[4].items():
    print(f'   {key}: {value}')

if adf_diff_result[1] <= 0.05:
    print("Conclusion: The differenced time series is likely stationary (reject H0).")
else:
    print("Conclusion: The differenced time series is likely non-stationary (fail to reject H0).")
```

**Reasoning**:
Since the differenced series is now stationary (d=1 confirmed by ADF test), I will now generate the Autocorrelation Function (ACF) and Partial Autocorrelation Function (PACF) plots for the `gdp_diff` series. These plots will help in visually identifying the appropriate 'p' (AR order) and 'q' (MA order) values for the ARIMA model by observing the lags where the correlation coefficients cut off or decay significantly.



```python
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

# Plot ACF and PACF for the differenced series
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

plot_acf(rajasthan_gdp_df['gdp_diff'].dropna(), lags=10, ax=axes[0])
axes[0].set_title('Autocorrelation Function (ACF) - Differenced GDP')

plot_pacf(rajasthan_gdp_df['gdp_diff'].dropna(), lags=10, ax=axes[1])
axes[1].set_title('Partial Autocorrelation Function (PACF) - Differenced GDP')

plt.tight_layout()
plt.show()
```

### Determining 'p' and 'q' for ARIMA Model from ACF and PACF Plots

**Analysis of ACF Plot (Autocorrelation Function):**

The ACF plot shows the correlation between the series and its lagged values. For the differenced GDP series (`gdp_diff`):

*   We observe a significant spike at **lag 1**. After lag 1, the correlations quickly fall within the confidence intervals (blue shaded area).
*   This pattern, a significant spike at lag 1 followed by a rapid decay or cutoff, suggests a potential **Moving Average (MA) order (q)** of **1**.

**Analysis of PACF Plot (Partial Autocorrelation Function):**

The PACF plot shows the partial correlation between the series and its lagged values, with the influence of the intermediate lags removed. For the differenced GDP series:

*   Similar to the ACF, we see a significant spike at **lag 1**. After lag 1, the partial correlations also quickly fall within the confidence intervals.
*   This pattern, a significant spike at lag 1 followed by a rapid decay or cutoff, suggests a potential **Autoregressive (AR) order (p)** of **1**.

**Conclusion for ARIMA Orders:**

*   From the Augmented Dickey-Fuller (ADF) test, we determined that the series required **one level of differencing (d=1)** to become stationary.
*   From the ACF plot, a significant spike at lag 1 suggests **q=1**.
*   From the PACF plot, a significant spike at lag 1 suggests **p=1**.

Therefore, a strong candidate for the ARIMA model orders (p, d, q) is likely **(1, 1, 1)**. It's common to test a few variations around these initial estimates, such as (1,1,0) or (0,1,1), to find the best-fitting model.

## Fit and Forecast with ARIMA Model

### Subtask:
Implement and fit an ARIMA model using the determined orders to Rajasthan's historical GDP data. Generate forecasts for future years and visualize these predictions alongside the historical data, including confidence intervals.


**Reasoning**:
I will import the `ARIMA` class, define the ARIMA order as (1, 1, 1) based on previous analysis (d=1, p=1, q=1 suggested by ACF/PACF), instantiate and fit the model to the original 'gdp' column, and then generate forecasts for 10 future years along with their confidence intervals. Finally, I will plot the historical GDP, forecasted GDP, and confidence intervals for clear visualization.



```python
from statsmodels.tsa.arima.model import ARIMA
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Define the ARIMA order (p, d, q) based on previous analysis
# d=1 was determined from the ADF test, p=1 and q=1 are initial guesses from ACF/PACF plots
arima_order = (1, 1, 1)

# Instantiate and fit the ARIMA model on the original 'gdp' series
# The differencing (d=1) will be handled internally by the ARIMA model
arima_model = ARIMA(rajasthan_gdp_df['gdp'], order=arima_order)
arima_results = arima_model.fit()

print("ARIMA model (1,1,1) fitted successfully.")

# Determine the number of steps for forecasting (10 years, consistent with linear regression)
last_historical_year = rajasthan_gdp_df.index.max().year
forecast_start_year = last_historical_year + 1
forecast_end_year = forecast_start_year + 9 # 10 years including the start year

# Generate future dates for forecasting
forecast_index = pd.to_datetime(pd.RangeIndex(start=forecast_start_year, stop=forecast_end_year + 1, name='year_id').map(str))

# Generate forecasts and confidence intervals
forecast_output = arima_results.get_forecast(steps=10)
forecast_gdp = forecast_output.predicted_mean
conf_int = forecast_output.conf_int()

# Create a DataFrame for forecasted GDP with confidence intervals
forecast_df = pd.DataFrame({
    'gdp': forecast_gdp,
    'lower_ci': conf_int.iloc[:, 0],
    'upper_ci': conf_int.iloc[:, 1]
}, index=forecast_index)


# Plot historical GDP, forecasted GDP, and confidence intervals
plt.figure(figsize=(14, 7))
sns.lineplot(x=rajasthan_gdp_df.index, y=rajasthan_gdp_df['gdp'], marker='o', label='Historical GDP')
sns.lineplot(x=forecast_df.index, y=forecast_df['gdp'], marker='x', linestyle='--', color='red', label='Forecasted GDP')
plt.fill_between(forecast_df.index, forecast_df['lower_ci'], forecast_df['upper_ci'], color='red', alpha=0.2, label='Confidence Interval')

plt.title("Rajasthan's GDP: Historical and ARIMA Forecast")
plt.xlabel('Year')
plt.ylabel('Real GDP')
plt.legend()
plt.grid(True)
plt.show()
```

## Determine SARIMA Orders (P, D, Q, S) and Fit Model

### Subtask:
Introduce the concept of seasonality in time series and explain how SARIMA (Seasonal ARIMA) extends ARIMA. Discuss how to identify seasonal orders (P, D, Q, S). Fit a SARIMA model to Rajasthan's GDP data, acknowledging that strong annual seasonality might not be present but demonstrating the model's structure.


### Understanding Seasonality and SARIMA Models

**Seasonality** in time series data refers to patterns that repeat over a fixed period. For example, retail sales might peak every December, or electricity consumption might rise during summer months each year. These recurring patterns, often driven by calendar events or natural cycles, can be captured and modeled to improve forecasting accuracy.

The **Seasonal ARIMA (SARIMA)** model is an extension of the AutoRegressive Integrated Moving Average (ARIMA) model, specifically designed to handle time series data that exhibits seasonal behavior. While ARIMA (p, d, q) models the non-seasonal components (autoregression, differencing, moving average), SARIMA adds a seasonal component, denoted as (P, D, Q, S):

*   **P (Seasonal AutoRegressive order):** The number of seasonal autoregressive terms.
*   **D (Seasonal Differencing order):** The number of seasonal differences required to make the series stationary seasonally.
*   **Q (Seasonal Moving Average order):** The number of seasonal moving average terms.
*   **S (Seasonal Period):** The number of time steps in a single seasonal period (e.g., 12 for monthly data with yearly seasonality, 4 for quarterly data with yearly seasonality).


**Determining Seasonal Orders for Annual Data (P, D, Q, S):**

For the Rajasthan GDP data, which is annual, we are observing a single data point per year. This means there isn't any 'within-year' seasonality to capture (like monthly or quarterly patterns within a 12-month cycle). Therefore, strong annual seasonality (e.g., a repeating pattern within a year) is generally not expected.

In such cases:
*   The **Seasonal Period (S)** is typically set to **1**, indicating no sub-annual seasonality.
*   The **Seasonal Differencing (D)** is set to **0** because there's no seasonal pattern to difference away over the period S=1.
*   The **Seasonal AR (P)** and **Seasonal MA (Q)** orders are also typically set to **0**, as there are no seasonal autoregressive or moving average components to model within a single annual period.

So, for annual data without explicit sub-annual seasonal patterns, the seasonal orders are usually (0, 0, 0, 1). This effectively means the SARIMA model simplifies to a regular ARIMA model when applied to annual data, as the seasonal component becomes trivial.

**Reasoning**:
Based on the instructions, I will now import the `SARIMAX` class, define the seasonal and non-seasonal orders, instantiate the SARIMA model with the specified orders for Rajasthan's GDP data, fit the model, and print a success message.



```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

# Non-seasonal ARIMA orders (p, d, q) determined previously
arima_order = (1, 1, 1)

# Seasonal orders (P, D, Q, S) for annual data
# S=1 indicates no sub-annual seasonality
# P, D, Q = 0 as no seasonal pattern is expected within S=1
sarimax_seasonal_order = (0, 0, 0, 1)

# Instantiate and fit the SARIMA model
# The differencing (d=1) will be handled internally by the SARIMA model
sarimax_model = SARIMAX(rajasthan_gdp_df['gdp'], order=arima_order, seasonal_order=sarimax_seasonal_order)
sarimax_results = sarimax_model.fit()

print("SARIMA model (1,1,1)(0,0,0,1) fitted successfully.")
```

**Reasoning**:

To resolve this while still explicitly defining the seasonal component as non-existent for annual data, I will change the seasonal period `S` from 1 to 0 in `sarimax_seasonal_order`.



```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

# Non-seasonal ARIMA orders (p, d, q) determined previously
arima_order = (1, 1, 1)

# Seasonal orders (P, D, Q, S) for annual data
# S=0 indicates no sub-annual seasonality (equivalent to (0,0,0) for P,D,Q)
# P, D, Q = 0 as no seasonal pattern is expected within S=0 (no seasonality)
sarimax_seasonal_order = (0, 0, 0, 0)

# Instantiate and fit the SARIMA model
# The differencing (d=1) will be handled internally by the SARIMA model
sarimax_model = SARIMAX(rajasthan_gdp_df['gdp'], order=arima_order, seasonal_order=sarimax_seasonal_order)
sarimax_results = sarimax_model.fit()

print("SARIMA model (1,1,1)(0,0,0,0) fitted successfully.")
```

## Explain ARIMA and SARIMA Forecasts for Layman

### Subtask:
Provide a simple, non-technical explanation of the ARIMA and SARIMA models, how their predictions are generated, and how they differ from simpler methods like linear regression. Focus on interpreting the forecasts and their implications for Rajasthan's economic outlook for a presentation.


### 1. Understanding the ARIMA Model (ARIMA: AutoRegressive Integrated Moving Average)

Think of the ARIMA model as a sophisticated crystal ball for time-based data, like our yearly GDP figures. It tries to predict future values by looking at three main things in the past:

*   **AR (AutoRegressive)**: This part looks at past GDP values themselves. If Rajasthan's GDP has consistently grown for the past few years, the AR component suggests it will likely continue to grow. It's like saying, "What happened yesterday often tells us something about what will happen today."
    *   `p` (the 'p' in ARIMA) tells us how many past GDP values the model considers.

*   **I (Integrated)**: Economic data, like GDP, often shows a continuous upward or downward trend, meaning it's not "stable" around a constant average. This 'I' component helps to make the data more stable by looking at the *differences* between consecutive GDP values, rather than the values themselves. For example, instead of looking at GDP in 2000, 2001, 2002, it looks at how much GDP *changed* from 2000 to 2001, and then from 2001 to 2002.
    *   `d` (the 'd' in ARIMA) tells us how many times we need to take these differences to make the data stable.

*   **MA (Moving Average)**: This part considers past prediction errors. If the model consistently underestimated GDP in the past, the MA component learns from those errors and adjusts future predictions. It's like saying, "If I kept making the same mistake, I should probably learn not to do that again."
    *   `q` (the 'q' in ARIMA) tells us how many past prediction errors the model considers.

In essence, ARIMA combines these three ideas: it uses past observations, accounts for underlying trends by differencing, and corrects for past prediction inaccuracies to forecast future GDP values. It's specifically designed for data where the order of observations matters, like a timeline.

### 2. Understanding the SARIMA Model (SARIMA: Seasonal AutoRegressive Integrated Moving Average)

The **SARIMA** model builds upon the ARIMA model, adding a special capability to handle patterns that repeat over fixed periods, known as **seasonality**. Imagine if we were looking at monthly GDP data; we might expect certain months (e.g., festival seasons) to consistently have higher GDP than others each year. This recurring pattern is seasonality.

SARIMA introduces four new components for seasonality, denoted as (P, D, Q, S):

*   **P (Seasonal AutoRegressive)**: Similar to the 'AR' part of ARIMA, but it looks at past values from the *same season* in previous cycles. For example, if we had monthly data, it would compare this January's GDP to last January's GDP.
*   **D (Seasonal Differencing)**: Similar to the 'I' part, but it removes seasonal trends. If there's a consistent boost in GDP every December, this component helps to make that seasonal pattern stable.
*   **Q (Seasonal Moving Average)**: Similar to the 'MA' part, but it accounts for past prediction errors from the *same season* in previous cycles.
*   **S (Seasonal Period)**: This is crucial; it defines the length of one seasonal cycle. For monthly data, S would be 12 (as the pattern repeats every 12 months). For quarterly data, S would be 4.

**SARIMA for Annual Data:**

Now, for our Rajasthan GDP data, we only have one data point per year. This means we don't have any 'within-year' patterns (like monthly or quarterly changes) to capture. In such cases, there isn't any meaningful seasonality to model *across seasons within a year*.

Therefore, when we apply SARIMA to annual data like this, the seasonal components (P, D, Q, S) are typically set to (0, 0, 0, 0) or (0, 0, 0, 1), which essentially tells the model: "There's no seasonality here, so just act like a regular ARIMA model." This means that for our annual GDP data, the SARIMA model will produce forecasts very similar, if not identical, to the ARIMA model because the seasonal component is effectively switched off.

### 3. Comparing ARIMA/SARIMA with Linear Regression

Previously, we used a **Linear Regression** model to forecast Rajasthan's GDP. Linear regression is a simpler model that looks for a straight-line relationship between a variable (like GDP) and another variable (like year). Its core assumption is that each data point is independent of the others, meaning the GDP in one year doesn't directly influence the GDP in the next year in a structured, time-dependent way.

However, economic data like GDP almost always has a strong **time dependency**. What happened last year *does* affect what happens this year. This is where **ARIMA** and **SARIMA** models shine.

**Key Differences:**

*   **Time Dependency**:
    *   **Linear Regression** treats each year's GDP as an independent observation. It looks for a general trend over time but doesn't explicitly model how the previous year's GDP or past errors influence the current one.
    *   **ARIMA/SARIMA** models are specifically designed for time series data. They explicitly capture the *sequential dependencies* of data points. This means they consider:
        *   How past GDP values directly influence future GDP values (AutoRegressive part).
        *   How past prediction errors influence future predictions (Moving Average part).
        *   How underlying trends need to be \"stabilized\" to make predictions more accurate (Integrated part).

*   **Seasonality**: While linear regression can sometimes be adapted to account for seasonality, SARIMA models are built from the ground up to recognize and incorporate repeating patterns over fixed periods.

In summary, while linear regression provides a useful initial forecast by finding a general trend, ARIMA and SARIMA offer a more nuanced and potentially more accurate prediction for time-dependent data like GDP by understanding the inherent structure and relationships within the historical sequence of observations.

**Understanding Confidence Intervals:**

The shaded areas around the forecasted lines (light red for ARIMA, light green for SARIMA) are called **Confidence Intervals**. These are very important because they tell us about the *uncertainty* in our predictions. Instead of giving us a single, exact number for future GDP, the models give us a range:

*   The **lower boundary** (bottom of the shaded area) represents the lower estimate of where GDP might be.
*   The **upper boundary** (top of the shaded area) represents the higher estimate of where GDP might be.

For example, if the forecast for a certain year is 260 units with a confidence interval of 220 to 300, it means we are, say, 95% confident that the actual GDP will fall somewhere within that 220-300 range. As we look further into the future, you'll see these shaded areas generally widen. This is expected, as predictions inherently become less certain the further out in time we try to forecast.
